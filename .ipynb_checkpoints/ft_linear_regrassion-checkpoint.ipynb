ALGORITMO DA TAXA DE ACTUALIZAÇÃO

```python
b0 = 0.0; b1 = 0.0
```

```python
par = []
par1 = []

x_tra = [par[0] for par in x_train]
y_tra = [par[1] for par in x_train]

x_train_transf = [par / 1000 for par in x_tra]
```

```python
print(x_tra[:5])
print(y_tra[:5])
print("-----------------------------------------")
print(x_train_transf[:5])
```

CALCULAR A PRIMEIRA PREVISÃO

```python
y_prev = []

for x in x_train_transf:
    prevision = b0 + b1 * x
    y_prev.append(prevision)
print(y_prev)
```

VAMOS CALCULAR O GRADIENTE

```python
m = len(x_tra)
db0 = 0; db1 = 0

for i in range(m):
    prevision = b0 + b1 * x_train_transf[i]
    error = prevision - y_tra[i]
    db0 += error
    db1 += error * x_train_transf[i]
db0 /= m
db1 /= m
```

```python
print(db0, db1)
```

VAMOS ACTUALIZAR OS PESOS

```python
alpha = 0.0000001
b0 = b0 - alpha * db0
b1 = b1 - alpha * db1
print(b0, b1)
```

```python
prevision = b0 + b1 * x_train_transf[0]
print("previsão = ", prevision, "valor esperado = ", y_tra[0])
```
